```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
trans = pd.read_csv("../data/raw/product_category_name_translation.csv")
```

```python
# First rows
trans.head()
```

```python
# Number of rows and columns
print("Shape:", trans.shape)

# Column names
print("Columns:", trans.columns.tolist())

# Data types and non-null counts
trans.info()

# Statistical summary
trans.describe()
```

```python

# Check duplicate rows
print("Duplicate rows:", trans.duplicated().sum())

# Check duplicate Portuguese categories
print("Duplicate Portuguese categories:",
      trans["product_category_name"].duplicated().sum())

# Check duplicate English categories
print("Duplicate English categories:",
      trans["product_category_name_english"].duplicated().sum())

# Check unnecessary leading/trailing spaces
print("Portuguese categories with extra spaces:",
      (trans["product_category_name"] !=
       trans["product_category_name"].str.strip()).sum())

print("English categories with extra spaces:",
      (trans["product_category_name_english"] !=
       trans["product_category_name_english"].str.strip()).sum())
```
