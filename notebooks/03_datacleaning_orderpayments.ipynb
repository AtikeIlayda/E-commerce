```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
order_payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
```

```python
# First rows
order_payments.head()
```

```python
# Number of rows and columns
print("Shape:", order_payments.shape)

# Column names
print("Columns:", order_payments.columns.tolist())

# Data types and non-null counts
order_payments.info()

# Statistical summary
order_payments.describe()
```

```python
#counts missing values in each column
order_payments.isnull().sum()
```

```python
#check for exact duplicate rows
order_payments.duplicated().sum()
```

```python
#explore the payment methods
print(order_payments["payment_type"].value_counts())

# Include percentages
print(
    (order_payments["payment_type"].value_counts(normalize=True) * 100)
    .round(2))
```

```python
display(
    order_payments[
        order_payments["payment_type"] == "not_defined"])
```

```python
#Investigate zero values
print(
    "Zero payment values:",
    (order_payments["payment_value"] == 0).sum()
)

print(
    "Zero installments:",
    (order_payments["payment_installments"] == 0).sum()
)
```

```python
zero_payments = order_payments[
    order_payments["payment_value"] == 0
]

display(zero_payments)
```

```python
# Get order IDs for zero-payment records
zero_payment_orders = order_payments.loc[
    order_payments["payment_value"] == 0,
    "order_id"
].unique()

# Display all payment records for those orders
zero_payment_details = order_payments[
    order_payments["order_id"].isin(zero_payment_orders)
].sort_values(["order_id", "payment_sequential"])

display(zero_payment_details)
```

```python
zero_installments = order_payments[
    order_payments["payment_installments"] == 0
]

display(zero_installments)
```

```python
#Investigate the highest payments
display(
    order_payments.nlargest(10, "payment_value")
)
```

```python
#Check orders with multiple payment records
payments_per_order = order_payments.groupby("order_id").size()

display(payments_per_order.describe())

print(
    "Orders with multiple payment records:",
    (payments_per_order > 1).sum()
)
```
