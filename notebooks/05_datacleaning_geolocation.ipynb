```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
geo = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")
```

```python
# First rows
geo.head()
```

```python
# Number of rows and columns
print("Shape:", geo.shape)

# Column names
print("Columns:", geo.columns.tolist())

# Data types and non-null counts
geo.info()

# Statistical summary
geo.describe()
```

```python
# Check coordinates outside broad Brazil bounds
outside_brazil = geo[
    ~geo["geolocation_lat"].between(-35, 6)
    | ~geo["geolocation_lng"].between(-75, -32)
]

print("Rows outside the broad Brazil bounds:", len(outside_brazil))

display(outside_brazil.head(20))
```

```python
geo["outside_brazil_bounds"] = (
    ~geo["geolocation_lat"].between(-35, 6)
    | ~geo["geolocation_lng"].between(-75, -32)
)

print(geo["outside_brazil_bounds"].value_counts())
```

```python
print("Exact duplicate rows:", geo.duplicated().sum())

print(
    "Unique postal code prefixes:",
    geo["geolocation_zip_code_prefix"].nunique()
)
```

```python
#counts missing values in each column
geo.isnull().sum()
```

```python
# Create a copy
geo_clean = geo.copy()

# Remove exact duplicate rows
geo_clean = geo_clean.drop_duplicates()

# Check the result
print("Original shape:", geo.shape)
print("Cleaned shape:", geo_clean.shape)
print("Remaining duplicates:", geo_clean.duplicated().sum())
```

```python
# Display suspicious coordinates
display(
    geo_clean[geo_clean["outside_brazil_bounds"]]
)
```

```python
geo_clean.to_csv("geolocation_clean.csv", index=False)
```
