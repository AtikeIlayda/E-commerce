```python
missing_summary = orders.isnull().sum()

missing_summary = missing_summary[missing_summary > 0]

print(missing_summary)
```

```python
orders["order_status"].value_counts()
```

```python
date_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

for column in date_columns:
    print(f"\nMissing values in {column}:")

    print(
        orders.loc[
            orders[column].isna(),
            "order_status"
        ].value_counts()
    )
```

```python
delivered_missing = orders[
    (orders["order_status"] == "delivered") &
    (
        orders["order_delivered_customer_date"].isna() |
        orders["order_delivered_carrier_date"].isna()
    )
]

print("Delivered orders missing a delivery milestone:",
      len(delivered_missing))

delivered_missing.head(10)
```

```python
for column in date_columns:
    print(f"\n--- {column} ---")

    summary = (
        orders.groupby("order_status")[column]
        .agg(
            total_orders="size",
            missing_values=lambda x: x.isna().sum()
        )
    )

    summary["missing_percent"] = (
        summary["missing_values"] /
        summary["total_orders"] * 100
    ).round(2)

    print(summary)
```

### Data Quality Check: Missing Values in Order Timestamps

**Dataset:** Olist Orders

**Columns investigated:**
- `order_approved_at`
- `order_delivered_carrier_date`
- `order_delivered_customer_date`

**Findings:**
- `order_approved_at`: 160 missing values (0.16%).
- `order_delivered_carrier_date`: 1,783 missing values (1.79%).
- `order_delivered_customer_date`: 2,965 missing values (2.98%).

The missing values were compared with `order_status`. Many appear consistent with orders that did not reach the corresponding stage of the purchasing or delivery process. However, some records marked as delivered are missing delivery timestamps and require further investigation.

**Cleaning decision:** Retain the missing values at this stage. Do not replace missing timestamps with estimated or average dates. Investigate delivered orders with incomplete timestamps and consider the availability of relevant dates when calculating delivery KPIs.

**Status:** Initial investigation completed; anomalous records under review.
