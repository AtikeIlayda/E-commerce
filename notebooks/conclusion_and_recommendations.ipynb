# Chapter 6 – Conclusion and Recommendations

## 6.1 Introduction

This chapter presents the overall conclusions of the study
"From Delivery to Loyalty: Understanding Customer Satisfaction
and Marketing Performance in E-Commerce".

The purpose of this chapter is to summarise the main findings,
answer the research questions, and draw conclusions from the
empirical analysis of the Olist Brazilian E-Commerce and
Marketing Funnel datasets.

Based on the findings presented and discussed in the previous
chapters, this chapter identifies practical recommendations
for e-commerce businesses and marketplace operators.

Finally, opportunities for future research are presented,
followed by the final remarks of the study.

## 6.2 Summary of the Study

This study investigated the relationship between operational
performance, customer satisfaction, and marketing performance
in e-commerce, using data from the Olist Brazilian E-Commerce
and Marketing Funnel datasets.

The research focused on six areas: delivery performance,
delivery speed, freight costs, product categories, marketing
funnel conversion, and the marketplace performance of
acquired sellers.

The analysis combined data preparation, data validation,
descriptive statistics, and inferential statistical methods.
The following statistical techniques were applied:

- Mann–Whitney U test to compare customer review scores
  between late and on-time or early deliveries.
- Spearman rank correlation to examine the relationships
  between delivery duration, freight costs, and customer
  review scores.
- Kruskal–Wallis test to examine differences in review
  score distributions across product categories.
- Chi-square test of independence to investigate the
  association between marketing lead origin and seller
  conversion.
- Descriptive statistics to evaluate the marketplace
  performance of acquired sellers.

The analysis was conducted using Python, SQL, PostgreSQL,
and Jupyter Notebook. The results were used to identify
patterns and associations between marketplace operations,
customer satisfaction, and seller acquisition performance.

Overall, the study adopted a data-driven approach to
understanding how operational and marketing activities
relate to performance in an e-commerce marketplace.

## 6.3 Answers to the Research Questions

The empirical findings provide answers to all six research
questions investigated in this study.

**RQ1 – Delivery Performance**

The analysis found a substantial difference in customer
satisfaction between late and on-time or early deliveries.
Late deliveries received an average review score of 2.57,
compared with 4.30 for on-time or early deliveries.
The Mann–Whitney U test confirmed that this difference
was statistically significant (p < .001).

Therefore, late delivery is strongly associated with
lower customer satisfaction.

**RQ2 – Delivery Speed**

The Spearman correlation analysis identified a negative
relationship between delivery duration and customer
satisfaction (ρ = -0.2353, p < .001).

This indicates that longer delivery times tend to be
associated with lower review scores. The relationship
is weaker than the difference observed between late
and on-time or early deliveries.

**RQ3 – Freight Costs**

The analysis identified a very weak negative association
between total freight costs and customer satisfaction
(ρ = -0.0881, p < .001).

Although statistically significant, the relationship
has limited practical strength. Higher freight costs
are associated with slightly lower review scores.

**RQ4 – Product Categories**

The Kruskal–Wallis test identified statistically significant
differences in review score distributions across product
categories (H = 817.70, p < .001).

This indicates that customer satisfaction varies across
product categories. The sensitivity analysis produced
a similar result when orders containing multiple
categories were excluded.

**RQ5 – Marketing Funnel Conversion Performance**

The chi-square test identified a statistically significant
association between lead origin and seller conversion
(χ²(9) = 137.687, p < .001).

Paid search achieved the highest conversion rate among
the identified marketing channels at 12.30%, while
organic search generated the largest number of converted
leads, with 271 conversions.

However, the association between lead origin and
conversion was weak (Cramér's V = 0.131).

**RQ6 – Acquired Seller Performance**

The descriptive analysis showed that 380 of the 842
acquired sellers could be matched to marketplace seller
records.

Across 4,405 eligible delivered orders, the order-weighted
average review score was 4.28, the average delivery
duration was 9.79 days, and the late delivery rate was
7.24%.

These results suggest generally positive observed
marketplace performance among matched acquired sellers.
However, the findings are descriptive and cannot be
generalised to all acquired sellers.

**Overall Answer to the Research Questions**

Taken together, the findings indicate that delivery
reliability and delivery speed are important factors
associated with customer satisfaction in the Olist
marketplace. Freight costs also show a negative
association, although its strength is very weak.
Customer satisfaction differs across product categories,
while marketing lead origin is associated with seller
conversion.

The acquired seller analysis further demonstrates the
importance of examining marketplace activity and
operational performance after seller acquisition.

Overall, the study shows that customer satisfaction
and marketing performance should be considered as
interconnected dimensions of e-commerce marketplace
performance, while recognising that the observed
relationships do not establish causation.

## 6.4 Overall Conclusion

The main objective of this study was to understand how
operational performance and marketing activities are
associated with customer satisfaction and marketplace
performance in e-commerce.

Based on the analysis of the Olist Brazilian E-Commerce
and Marketing Funnel datasets, the findings demonstrate
that several operational and marketing-related factors
are associated with marketplace outcomes.

The most notable finding concerns delivery performance.
Orders delivered late received substantially lower
customer review scores than orders delivered on time
or early. In addition, longer delivery durations were
associated with lower customer satisfaction. These
results highlight the importance of reliable and timely
delivery in the e-commerce customer experience.

Freight costs showed a statistically significant but
very weak negative association with customer review
scores. Customer satisfaction also differed across
product categories, suggesting that the customer
experience varies depending on the type of product
purchased.

Regarding marketing performance, lead origin was
significantly associated with seller conversion.
However, the strength of this relationship was weak,
indicating that acquisition channel alone does not
fully explain conversion outcomes.

The acquired seller analysis further demonstrated
that matched sellers generated measurable marketplace
activity and generally positive customer review scores
and delivery performance. Nevertheless, incomplete
seller matching limits the extent to which these
findings can be generalised.

Overall, the study suggests that successful e-commerce
marketplace management requires attention to both
customer-facing operational performance and marketing
activities. Attracting sellers is important, but their
subsequent marketplace activity and operational
performance also matter. Similarly, reliable delivery
and effective fulfilment processes are important
considerations for maintaining customer satisfaction.

The study therefore concludes that delivery reliability,
delivery speed, product category, freight costs, and
marketing lead origin are relevant factors associated
with customer and marketplace outcomes in the Olist
context. These findings provide an empirical basis
for understanding the connection between marketing
activities, marketplace operations, and customer
satisfaction, while recognising the limitations of
observational data.

## 6.5 Recommendations

Based on the findings of this study, several recommendations
can be made for e-commerce businesses and marketplace operators.

**1. Prioritise delivery reliability**

Businesses should prioritise meeting promised delivery dates.
The substantial difference in customer review scores between
late and on-time or early deliveries demonstrates the importance
of delivery reliability.

Recommended actions include:
- Monitoring late delivery rates regularly.
- Improving coordination with logistics providers.
- Identifying recurring causes of delivery delays.
- Providing customers with realistic delivery estimates.
- Communicating proactively when delays occur.

**2. Improve delivery speed and fulfilment efficiency**

Businesses should review their order processing and fulfilment
activities to identify opportunities to reduce unnecessary
delivery time.

Recommended actions include:
- Monitoring average delivery duration.
- Improving warehouse and order-processing efficiency.
- Reviewing logistics routes and carrier performance.
- Identifying bottlenecks in the fulfilment process.

**3. Ensure transparent and competitive freight pricing**

Although freight costs showed only a very weak association
with customer satisfaction, shipping charges should remain
an important consideration in pricing decisions.

Recommended actions include:
- Displaying delivery charges clearly before checkout.
- Reviewing freight pricing regularly.
- Considering free-shipping thresholds where financially
  appropriate.
- Offering different delivery options to customers.

**4. Monitor customer satisfaction by product category**

The differences in review score distributions across
product categories indicate that businesses should assess
customer satisfaction at category level.

Recommended actions include:
- Identifying categories with comparatively low review scores.
- Monitoring product quality and return patterns.
- Reviewing seller performance within individual categories.
- Investigating category-specific delivery and packaging issues.

**5. Optimise marketing channel performance**

The marketing funnel analysis indicates that lead origin
is associated with seller conversion. However, conversion
rates and conversion volumes provide different insights.

Recommended actions include:
- Monitoring conversion rates by acquisition channel.
- Comparing conversion volumes across channels.
- Evaluating marketing expenditure and acquisition costs.
- Assessing the long-term marketplace performance of acquired
  sellers rather than focusing only on initial conversions.

**6. Strengthen acquired seller onboarding and monitoring**

The acquired seller analysis highlights the importance
of evaluating sellers after their acquisition.

Recommended actions include:
- Providing structured onboarding support.
- Monitoring seller order activity and customer reviews.
- Tracking delivery performance over time.
- Identifying sellers who may need additional operational
  assistance.
- Providing category-specific guidance where necessary.

**7. Develop an integrated performance monitoring framework**

Marketplace operators should monitor marketing and
operational performance together rather than treating
them as separate activities.

A useful performance framework could include:

| Performance Area | Suggested KPI |
|---|---|
| Customer satisfaction | Average review score |
| Delivery reliability | Late delivery rate |
| Delivery speed | Average delivery duration |
| Shipping costs | Average freight cost per order |
| Product performance | Average review score by category |
| Marketing effectiveness | Lead conversion rate |
| Seller acquisition | Number of acquired sellers |
| Seller performance | Orders per acquired seller |

Overall, the recommendations emphasise the importance
of combining effective seller acquisition with reliable
marketplace operations and continuous customer
satisfaction monitoring.

## 6.6 Future Research

Although this study provides useful insights into customer
satisfaction and marketing performance in e-commerce,
several opportunities exist for further research.

**1. Investigate additional factors influencing customer satisfaction**

Future studies could examine additional variables such as
product quality, customer expectations, payment methods,
return rates, customer service, and order value. Including
these factors could provide a more comprehensive understanding
of the determinants of customer satisfaction.

**2. Apply multivariate statistical methods**

This study primarily used non-parametric statistical tests
and descriptive analysis. Future research could apply
multivariate methods, such as multiple regression or
appropriate ordinal regression models, to examine the
relationships between several explanatory variables and
customer satisfaction simultaneously.

Such methods could also help control for potential
confounding factors.

**3. Examine differences across geographical regions**

Future research could investigate whether delivery
performance and customer satisfaction differ between
geographical regions. This could help identify regional
logistics challenges and differences in customer expectations.

**4. Investigate marketing acquisition costs and seller value**

Further research could connect marketing funnel data
with marketing expenditure and long-term seller activity.
This would allow researchers to examine customer acquisition
costs, seller retention, and the long-term value generated
by acquired sellers.

**5. Extend the observation period**

A longer observation period could provide additional insight
into seller development and marketplace performance.
It could help determine whether newly acquired sellers
increase their activity over time and whether their
operational performance improves with experience.

**6. Improve seller matching and performance measurement**

Future studies could investigate more comprehensive
methods of linking acquired sellers with marketplace
records. This would reduce the limitations associated
with unmatched sellers and provide a more complete
assessment of seller acquisition outcomes.

**7. Conduct comparative studies across marketplaces**

Finally, future research could compare Olist with other
e-commerce marketplaces or datasets from different
countries. Such comparisons could help establish whether
the relationships observed in this study are specific
to the Brazilian marketplace or also occur in other
e-commerce environments.

Overall, future research could build on this study by
using broader datasets, more comprehensive performance
measures, longer observation periods, and advanced
statistical methods. This would contribute to a deeper
understanding of how marketing activities and operational
performance relate to customer satisfaction and
marketplace success.

## 6.7 Final Remarks

This study examined the relationship between delivery
performance, customer satisfaction, and marketing
performance in the context of e-commerce, using the
Olist Brazilian E-Commerce and Marketing Funnel datasets.

The findings demonstrate that reliable delivery is
particularly important in relation to customer satisfaction.
Late deliveries were associated with substantially lower
review scores, while longer delivery durations were also
associated with lower customer satisfaction.

The study also identified differences in customer
satisfaction across product categories and a weak
association between freight costs and review scores.
Furthermore, marketing lead origin was associated
with seller conversion, while the acquired seller
analysis provided insights into the marketplace activity
and operational performance of matched sellers.

The overall findings highlight that e-commerce
marketplace performance involves more than attracting
customers or acquiring sellers. Reliable logistics,
effective fulfilment, transparent pricing, seller support,
and continuous performance monitoring are also
important considerations.

Although the study is based on observational data
from a specific marketplace and therefore has
limitations in terms of generalisability and causal
interpretation, it provides useful empirical insights
into the relationship between marketing activities,
marketplace operations, and customer satisfaction.

Ultimately, the study emphasises that sustainable
e-commerce marketplace performance requires an
integrated approach in which marketing acquisition
and operational service quality work together to
support a positive customer experience.
