```python
# Convert order date columns from strings to datetime format

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

# Check the updated data types
print(orders[date_columns].dtypes)
```

```python
# Check the earliest and latest dates in each column

for col in date_columns:
    print(f"\n{col}")
    print("Earliest date:", orders[col].min())
    print("Latest date:", orders[col].max())
    print("Missing dates:", orders[col].isna().sum())
```

```python
# Check the chronological order of order events

print(
    "Approval before purchase:",
    (orders["order_approved_at"] < orders["order_purchase_timestamp"]).sum()
)

print(
    "Carrier date before purchase:",
    (orders["order_delivered_carrier_date"] < orders["order_purchase_timestamp"]).sum()
)

print(
    "Customer delivery before carrier date:",
    (orders["order_delivered_customer_date"] < orders["order_delivered_carrier_date"]).sum()
)

print(
    "Delivery after estimated date:",
    (orders["order_delivered_customer_date"] > orders["order_estimated_delivery_date"]).sum()
)
```

```python
# Inspect orders where carrier handover occurred before purchase

carrier_before_purchase = orders[
    orders["order_delivered_carrier_date"]
    < orders["order_purchase_timestamp"]
]

print("Number of affected orders:", len(carrier_before_purchase))

print(
    carrier_before_purchase[
        [
            "order_id",
            "order_status",
            "order_purchase_timestamp",
            "order_approved_at",
            "order_delivered_carrier_date"
        ]
    ].head(10).to_string(index=False)
)
```

```python
# Calculate the time difference between purchase and carrier handover

carrier_before_purchase = orders[
    orders["order_delivered_carrier_date"]
    < orders["order_purchase_timestamp"]
].copy()

carrier_before_purchase["difference_minutes"] = (
    carrier_before_purchase["order_purchase_timestamp"]
    - carrier_before_purchase["order_delivered_carrier_date"]
).dt.total_seconds() / 60

print("Number of affected orders:", len(carrier_before_purchase))

print("\nDifference in minutes:")
print(carrier_before_purchase["difference_minutes"].describe())

print("\nLargest time differences:")
print(
    carrier_before_purchase[
        [
            "order_id",
            "order_purchase_timestamp",
            "order_delivered_carrier_date",
            "difference_minutes"
        ]
    ]
    .sort_values("difference_minutes", ascending=False)
    .head(10)
    .to_string(index=False)
)
```

```python
# Classify carrier timestamps recorded before purchase

conditions = [
    carrier_before_purchase["difference_minutes"] <= 60,
    (carrier_before_purchase["difference_minutes"] > 60)
    & (carrier_before_purchase["difference_minutes"] <= 1440),
    carrier_before_purchase["difference_minutes"] > 1440
]

labels = [
    "Up to 1 hour",
    "1 to 24 hours",
    "More than 24 hours"
]

carrier_before_purchase["difference_category"] = pd.Series(
    "Unclassified",
    index=carrier_before_purchase.index
)

for condition, label in zip(conditions, labels):
    carrier_before_purchase.loc[condition, "difference_category"] = label

print(
    carrier_before_purchase["difference_category"]
    .value_counts()
    .reindex(labels, fill_value=0)
)
```

```python
# Inspect orders where customer delivery precedes carrier handover

delivery_before_carrier = orders[
    orders["order_delivered_customer_date"]
    < orders["order_delivered_carrier_date"]
].copy()

delivery_before_carrier["difference_hours"] = (
    delivery_before_carrier["order_delivered_carrier_date"]
    - delivery_before_carrier["order_delivered_customer_date"]
).dt.total_seconds() / 3600

print("Number of affected orders:", len(delivery_before_carrier))

print("\nDifference in hours:")
print(delivery_before_carrier["difference_hours"].describe())

print("\nAffected orders:")
print(
    delivery_before_carrier[
        [
            "order_id",
            "order_status",
            "order_delivered_carrier_date",
            "order_delivered_customer_date",
            "difference_hours"
        ]
    ]
    .sort_values("difference_hours", ascending=False)
    .head(10)
    .to_string(index=False)
)
```

```python
# Compare actual and estimated delivery calendar dates

actual_date = orders["order_delivered_customer_date"].dt.normalize()
estimated_date = orders["order_estimated_delivery_date"].dt.normalize()

valid = actual_date.notna() & estimated_date.notna()

late = valid & (actual_date > estimated_date)
on_time = valid & (actual_date == estimated_date)
early = valid & (actual_date < estimated_date)

print("Orders with actual delivery dates:", valid.sum())
print("Delivered late:", late.sum())
print("Delivered on estimated date:", on_time.sum())
print("Delivered early:", early.sum())

print(
    "Late delivery percentage:",
    round(late.sum() / valid.sum() * 100, 2), "%"
)
```


# Step 1: Data Audit and Initial Exploration — Orders Dataset

## 1. Objective

The objective of this step was to inspect the orders dataset and assess its data quality before further analysis. The audit included checking the dataset structure, data types, missing values, date ranges, chronological consistency, and delivery performance.

Python and Pandas were used to load and inspect the data.

## 2. Dataset Overview

- **Number of rows:** 99,441
- **Number of columns:** 8
- **Date columns:** 5
- **Purchase date range:** September 2016 – October 2018
- **Duplicate rows:** Not yet recorded

The five date columns were converted from strings to datetime format.

## 3. Missing Values

| Column | Missing Values | Assessment |
|---|---:|---|
| `order_id` | 0 | Complete |
| `customer_id` | 0 | Complete |
| `order_status` | 0 | Complete |
| `order_purchase_timestamp` | 0 | Complete |
| `order_approved_at` | 160 | Investigate |
| `order_delivered_carrier_date` | 1,783 | Investigate |
| `order_delivered_customer_date` | 2,965 | Investigate |
| `order_estimated_delivery_date` | 0 | Complete |

Missing timestamps were examined in relation to order status. Some missing values may be explained by canceled, unavailable, or uncompleted orders. However, missing timestamps were also found in orders marked as delivered.

No missing values were removed during this audit.

## 4. Date Ranges

| Date Column | Earliest Date | Latest Date | Missing |
|---|---|---|---:|
| `order_purchase_timestamp` | 2016-09-04 | 2018-10-17 | 0 |
| `order_approved_at` | 2016-09-15 | 2018-09-03 | 160 |
| `order_delivered_carrier_date` | 2016-10-08 | 2018-09-11 | 1,783 |
| `order_delivered_customer_date` | 2016-10-11 | 2018-10-17 | 2,965 |
| `order_estimated_delivery_date` | 2016-09-30 | 2018-11-12 | 0 |

The date ranges were reviewed to identify potential irregularities and prepare the dataset for time-based analysis.

## 5. Chronological Consistency

The expected order sequence is generally:

1. Purchase
2. Approval
3. Carrier handover
4. Customer delivery

The following potential inconsistencies were identified:

| Check | Records | Finding |
|---|---:|---|
| Approval before purchase | 0 | No violations found |
| Carrier handover before purchase | 166 | Potential inconsistencies |
| Customer delivery before carrier handover | 23 | Potential inconsistencies |

### Carrier Timestamp Findings

Of the 166 orders with carrier timestamps earlier than purchase timestamps:

| Difference | Number of Orders |
|---|---:|
| Up to 1 hour | 121 |
| 1–24 hours | 43 |
| More than 24 hours | 2 |
| **Total** | **166** |

The largest difference was approximately 171 days.

### Customer Delivery Timestamp Findings

For the 23 orders with customer delivery timestamps earlier than carrier handover:

- Median discrepancy: approximately 39.9 hours.
- Maximum discrepancy: approximately 386.3 hours.

These records were flagged for investigation. No timestamps were modified because the reasons for the inconsistencies could not be confirmed.

## 6. Delivery Performance

Actual customer delivery dates were compared with estimated delivery dates using calendar dates.

A total of 96,476 orders had recorded customer delivery dates.

| Delivery Outcome | Number of Orders | Percentage |
|---|---:|---:|
| Delivered early | 88,649 | 91.9% |
| Delivered on estimated date | 1,292 | 1.3% |
| Delivered late | 6,535 | 6.77% |
| **Total** | **96,476** | **100%** |

Most orders with recorded delivery dates arrived before the estimated date. Approximately 6.77% arrived after the estimated calendar date.

These results are preliminary because the identified timestamp inconsistencies may affect delivery-performance calculations.

## 7. Main Findings

- The dataset contains 99,441 rows and 8 columns.
- Five date columns were converted to datetime format.
- Missing values were found in approval and delivery timestamps.
- A total of 166 potential carrier timestamp inconsistencies were identified.
- A total of 23 potential customer delivery timestamp inconsistencies were identified.
- The preliminary late delivery rate was 6.77% among orders with recorded customer delivery dates.
- No records or timestamps were deleted or modified.

## 8. Conclusion

The initial audit provided an overview of the dataset structure, missing values, date ranges, chronological consistency, and delivery performance.

The dataset can proceed to further exploratory analysis, but the identified timestamp inconsistencies should be investigated before drawing final conclusions.
