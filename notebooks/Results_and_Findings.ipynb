## 4.2 RQ1 – Delivery Performance and Customer Satisfaction

The first research question investigated whether delivery performance is associated with customer satisfaction in the Olist e-commerce marketplace. In particular, the analysis examined whether customers who experienced late deliveries provided different review scores compared with customers whose orders arrived on time or earlier than estimated.

The analysis included 95,178 delivered orders with valid delivery chronology and available customer review scores. Of these, 7,609 orders (7.99%) were classified as late, while 87,569 orders (92.01%) were delivered on time or earlier.

Descriptive analysis revealed a substantial difference in customer review scores between the two groups. Late deliveries received an average review score of 2.57, compared with 4.30 for orders delivered on time or earlier. The median review score was 2 for late deliveries and 5 for on-time or early deliveries.

To determine whether this difference was statistically significant, a Mann–Whitney U test was conducted. The results indicated a statistically significant difference in review-score distributions between the two delivery-performance groups (U = 148,395,758, p < .001). The rank-biserial correlation was −0.555, indicating a substantial difference between the groups.

Consequently, the null hypothesis was rejected in favour of the alternative hypothesis. The findings indicate that late delivery is associated with considerably lower customer satisfaction in the Olist marketplace.

Overall, the results highlight the importance of delivery reliability in the e-commerce customer experience. However, the findings demonstrate an association rather than a causal relationship.

## 4.3 RQ2 – Delivery Speed and Customer Satisfaction

The second research question examined the relationship between delivery speed and customer satisfaction in the Olist e-commerce marketplace. Unlike the previous analysis, which focused on whether orders arrived late or on time, this analysis investigated whether the actual delivery duration was associated with customer review scores.

The analysis included 95,178 delivered orders with valid delivery chronology and available customer review scores. The average delivery duration was 12.52 days (SD = 9.47), with a median of 10.21 days. Delivery durations ranged from 0.53 to 208.35 days.

Spearman's rank correlation was used to examine the relationship between delivery duration and customer review scores. This non-parametric method was selected because it assesses monotonic relationships without requiring the variables to follow a normal distribution.

The results revealed a statistically significant negative correlation between delivery duration and customer review scores (ρ = −0.2353, p < .001). This indicates that longer delivery durations tended to be associated with lower customer satisfaction. However, the strength of the relationship was weak to moderate.

A sensitivity analysis was also conducted to examine the influence of unusually long delivery durations. When orders exceeding 60 days were excluded, the correlation remained negative and statistically significant (ρ = −0.2306). When the analysis was restricted to deliveries of 30 days or less, the correlation was weaker but remained negative (ρ = −0.1595). These results indicate that the negative relationship was not solely attributable to extremely long delivery durations.

Consequently, the null hypothesis was rejected in favour of the alternative hypothesis. The findings support the existence of a negative monotonic association between delivery duration and customer satisfaction.

Overall, the results suggest that faster delivery is associated with higher customer satisfaction in the Olist marketplace. Nevertheless, delivery speed alone does not explain all variations in customer reviews, and the findings do not establish a causal relationship.

## 4.4 RQ3 – Freight Costs and Customer Satisfaction

The third research question investigated the relationship between freight costs and customer satisfaction in the Olist e-commerce marketplace. The analysis examined whether orders with higher freight costs tended to receive different customer review scores.

The analysis included 95,178 delivered orders with valid delivery chronology and available freight-cost and customer review data. The average total freight cost per order was 22.76, with a standard deviation of 21.55. The median freight cost was 17.16.

Spearman's rank correlation was applied to examine the relationship between total freight costs and customer review scores. This non-parametric method was selected because it assesses monotonic associations between variables without requiring a normal distribution.

The results revealed a statistically significant but very weak negative correlation between freight costs and customer review scores (ρ = −0.0881, p < .001). This indicates that higher freight costs tended to be associated with slightly lower customer satisfaction. However, the small correlation coefficient suggests that the practical strength of this relationship was limited.

To further explore this relationship, orders were divided into four approximately equal groups based on their total freight costs. The lowest freight-cost quartile recorded the highest average review score (M = 4.316), whereas the highest freight-cost quartile recorded the lowest average score (M = 3.944). This represents a difference of 0.372 points on the five-point review scale. The median review score remained 5 across all four groups, suggesting that lower ratings contributed to the differences in average scores.

Consequently, the null hypothesis was rejected in favour of the alternative hypothesis. Nevertheless, the very weak correlation indicates that freight costs have only a limited association with customer satisfaction.

Overall, the findings suggest that higher freight costs may be associated with slightly lower customer satisfaction, although other aspects of the purchasing and delivery experience are likely to play a more substantial role. The results demonstrate an association rather than a causal relationship.

## 4.5 RQ4 – Product Categories and Customer Satisfaction

The fourth research question investigated whether customer satisfaction differs across product categories in the Olist e-commerce marketplace. The analysis examined variations in customer review scores across the different product categories represented in the dataset.

The initial dataset contained 95,944 order-category observations across 74 product categories. To ensure a reasonable number of observations for each category, only categories associated with at least 50 orders were included in the main statistical analysis. This resulted in 59 eligible categories and 95,620 order-category observations.

A Kruskal–Wallis test was conducted to determine whether customer review-score distributions differed across the product categories. This non-parametric method was selected because it allows the comparison of multiple independent groups without requiring normally distributed data.

The results indicated statistically significant differences in review-score distributions across the 59 product categories (H = 817.70, p < .001). Consequently, the null hypothesis was rejected, indicating that at least one product category had a different customer review-score distribution.

To identify which categories differed, Dunn's post-hoc pairwise comparisons with Holm correction were conducted. Of the 1,711 pairwise comparisons, 202 remained statistically significant after adjustment for multiple testing. One of the largest differences was observed between office furniture (M = 3.644) and books general interest (M = 4.534), representing a difference of 0.890 points on the five-point review scale.

A sensitivity analysis was conducted to address the potential dependence between observations from orders containing products from multiple categories. The analysis was repeated using only orders associated with exactly one product category. This resulted in 94,134 observations across 58 categories meeting the minimum threshold of 50 orders. The Kruskal–Wallis test remained statistically significant (H = 778.40, p < .001), while 181 of the 1,653 Dunn pairwise comparisons remained significant after Holm correction. The largest difference continued to be between office furniture and books general interest.

The consistency of the findings across the main and sensitivity analyses provides further support for differences in customer satisfaction across product categories. Therefore, the alternative hypothesis was supported.

Overall, the findings suggest that product category is associated with customer satisfaction in the Olist marketplace. However, the results do not establish a causal relationship, and the Kruskal–Wallis test evaluates differences in distributions rather than differences in means.

## 4.6 RQ5 – Marketing Funnel Conversion Performance

The fifth research question investigated the effectiveness of the Olist marketing funnel in converting marketing qualified leads into acquired sellers and examined whether conversion performance differed according to lead origin.

The analysis included 8,000 marketing qualified leads, of which 842 were successfully converted into acquired sellers. This represented an overall conversion rate of 10.53%, while the remaining 7,158 leads (89.47%) did not have a corresponding closed-deal record.

Conversion performance varied across marketing origins. Leads with an unknown origin recorded the highest conversion rate at 16.65%. Among identifiable marketing channels, paid search achieved the highest conversion rate (12.30%), followed by organic search (11.80%) and direct traffic (11.22%). Organic search generated the largest absolute number of acquired sellers, with 271 conversions, followed by paid search with 195 conversions. In contrast, email (3.04%) and other sources (2.67%) recorded the lowest conversion rates.

A chi-square test of independence was conducted to determine whether lead origin was associated with conversion status. The results revealed a statistically significant association between the two variables (χ²(9) = 137.687, p < .001). The minimum expected cell frequency was 6.84, satisfying the expected-frequency assumption of the test. Cramér's V was 0.1312, indicating a relatively weak association.

Consequently, the null hypothesis was rejected in favour of the alternative hypothesis. The findings indicate that lead origin and conversion status are not independent. Nevertheless, the relatively small effect size suggests that marketing origin alone has limited explanatory power in relation to lead conversion.

Overall, the findings demonstrate that marketing origin is associated with the effectiveness of the Olist lead conversion process. Organic search was the largest contributor to acquired sellers in terms of conversion volume, whereas paid search achieved the highest conversion rate among identifiable channels. These findings highlight the importance of evaluating marketing performance using both conversion rates and the absolute number of acquired sellers. However, the results indicate an association rather than a causal relationship.

## 4.7 RQ6 – Marketplace Performance of Acquired Sellers

The sixth research question investigated the marketplace performance of sellers acquired through the Olist marketing funnel, focusing on sales activity, customer satisfaction and delivery performance.

The analysis included 842 acquired sellers. Of these, 380 sellers (45.13%) could be matched to marketplace seller records, while 462 sellers (54.87%) could not be matched. The matched sellers generated 4,518 marketplace orders, comprising 5,044 items sold and a total product value of approximately R$676,851.46.

Among the matched sellers, 376 had available customer satisfaction and delivery performance data. These sellers accounted for 4,405 eligible delivered orders. Using an order-weighted approach, the average customer review score was 4.28 on a five-point scale. The average delivery duration was 9.79 days, while the weighted late delivery rate was 7.24%. These results indicate generally positive customer satisfaction and relatively reliable delivery performance among the acquired sellers that could be linked to marketplace activity.

Further analysis examined differences in customer satisfaction across seller business segments. Sports and leisure recorded an average review score of 4.70, followed by bags and backpacks (4.62) and health and beauty (4.58). Health and beauty was particularly notable because it represented 45 matched sellers and 705 marketplace orders. In contrast, home office furniture recorded the lowest average review score (3.39), followed by phone and mobile (3.44). However, these lower averages were based on only three and five sellers, respectively, and should therefore be interpreted cautiously.

Differences were also observed in delivery performance across business segments. Sports and leisure recorded a relatively low average late delivery rate of 1.62%, whereas home decor recorded 13.78%. The watches segment had a late delivery rate of 16.84%, although this result was based on only three sellers.

Overall, the findings suggest that acquired sellers demonstrated generally positive marketplace performance in terms of customer satisfaction and delivery reliability. Nevertheless, performance varied across business segments, and the analysis was limited by the fact that only 45.13% of acquired sellers could be matched to marketplace records.

As the analysis was based on descriptive statistics rather than inferential hypothesis testing, the findings provide descriptive support for the alternative hypothesis but do not allow the null hypothesis to be formally rejected. Furthermore, the results do not establish that marketing acquisition caused subsequent seller performance.

## 4.8 Summary of Findings and Hypothesis Decisions

This chapter presented the empirical findings for the six research questions investigating the relationships between delivery performance, customer satisfaction, product characteristics, marketing funnel conversion and acquired seller performance in the Olist e-commerce marketplace.

The findings demonstrate that delivery performance and delivery speed are associated with customer satisfaction. Late deliveries were associated with substantially lower customer review scores, while longer delivery durations showed a negative relationship with customer satisfaction. Freight costs also demonstrated a statistically significant, although very weak, negative association with review scores.

Furthermore, customer satisfaction differed significantly across product categories, indicating that the type of product purchased was associated with variations in customer review scores. The marketing funnel analysis also identified a statistically significant association between lead origin and conversion status, although the strength of this relationship was relatively weak.

Finally, the acquired seller analysis showed generally positive customer satisfaction and delivery performance among sellers successfully matched to marketplace records. However, performance varied across business segments, and the matching limitations prevented the findings from being generalised to all acquired sellers.

Table 4.1 summarises the statistical methods, principal findings and hypothesis decisions for all six research questions.

Overall, the results suggest that customer satisfaction and marketing performance in e-commerce are associated with multiple operational and marketing factors. The findings highlight the importance of delivery reliability, delivery speed, freight costs, product category and lead origin, while also demonstrating the need to monitor seller performance after acquisition. The following chapter discusses these findings in relation to the relevant literature and theoretical framework.
