```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")
```

```python
# First rows
sellers.head()
```

```python
# Number of rows and columns
print("Shape:", sellers.shape)

# Column names
print("Columns:", sellers.columns.tolist())

# Data types and non-null counts
sellers.info()

# Statistical summary
sellers.describe()
```

```python

# 1. Check missing values
print("Missing values:")
print(sellers.isnull().sum())

# 2. Check duplicate rows
print("\nFully duplicated rows:", sellers.duplicated().sum())

# 3. Check duplicate seller IDs
print("Duplicate seller IDs:", sellers["seller_id"].duplicated().sum())

# 4. Check ZIP code prefix formatting and range
print("\nZIP code prefix summary:")
print(sellers["seller_zip_code_prefix"].describe())

print("\nZIP prefixes with fewer than 5 digits:")
print(
    sellers[
        sellers["seller_zip_code_prefix"].astype(str).str.len() < 5
    ]["seller_zip_code_prefix"].head(10)
)

# 5. Check unique cities and states
print("\nUnique cities:", sellers["seller_city"].nunique())
print("Unique states:", sellers["seller_state"].nunique())

print("\nSeller state counts:")
print(sellers["seller_state"].value_counts())

# 6. Check text formatting
print("\nBlank seller cities:",
      sellers["seller_city"].str.strip().eq("").sum())

print("Blank seller states:",
      sellers["seller_state"].str.strip().eq("").sum())
```

```python

# Create a copy for cleaning
sellers_cleaned = sellers.copy()

# 1. Convert ZIP code prefixes to strings
# and ensure they have five digits
sellers_cleaned["seller_zip_code_prefix"] = (
    sellers_cleaned["seller_zip_code_prefix"]
    .astype(str)
    .str.zfill(5)
)

# 2. Standardize city names
sellers_cleaned["seller_city"] = (
    sellers_cleaned["seller_city"].str.strip()
)

# 3. Standardize state abbreviations
sellers_cleaned["seller_state"] = (
    sellers_cleaned["seller_state"].str.strip().str.upper()
)

# 4. Check the results
print("Original shape:", sellers.shape)
print("Cleaned shape:", sellers_cleaned.shape)

print("\nUpdated data types:")
print(sellers_cleaned.dtypes)

print("\nMissing values:")
print(sellers_cleaned.isnull().sum())

print("\nDuplicate seller IDs:",
      sellers_cleaned["seller_id"].duplicated().sum())

print("\nZIP code prefix examples:")
print(sellers_cleaned["seller_zip_code_prefix"].head(10))

print("\nPreview:")
print(sellers_cleaned.head())
```

```python
sellers_cleaned.to_csv("sellers_cleaned.csv", index=False)

print("Sellers dataset saved successfully!")
```

Sellers Dataset – Data Cleaning Summary

The Olist sellers dataset originally contained 3,095 rows and 4 columns.

Cleaning Steps

Checked for missing values; none were found.
Checked for duplicate rows and duplicate seller IDs; none were found.
Checked city and state columns for blank values; none were found.
Converted seller_zip_code_prefix from integer to string to preserve its identifier format.
Used .str.zfill(5) to ensure all ZIP code prefixes contain five digits.
Used .str.strip() to remove unnecessary leading and trailing spaces from city and state names.
Standardized state abbreviations to uppercase.
Final Result

The cleaned dataset retains all 3,095 rows and 4 columns. No seller records were deleted. The cleaned version is saved as sellers_cleaned.csv, while the original dataset remains unchanged.
