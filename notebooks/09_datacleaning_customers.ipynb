```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
```

```python
# First rows
customers.head()
```

```python
# Number of rows and columns
print("Shape:", customers.shape)

# Column names
print("Columns:", customers.columns.tolist())

# Data types and non-null counts
customers.info()

# Statistical summary
customers.describe()
```

```python

# 1. Check missing values
print("Missing values:")
print(customers.isnull().sum())

# 2. Check duplicate rows
print("\nFully duplicated rows:", customers.duplicated().sum())

# 3. Check duplicate customer IDs
print("Duplicate customer IDs:",
      customers["customer_id"].duplicated().sum())

# 4. Check duplicate unique customer IDs
print("Duplicate unique customer IDs:",
      customers["customer_unique_id"].duplicated().sum())

# 5. Check ZIP code prefixes
print("\nZIP code prefix summary:")
print(customers["customer_zip_code_prefix"].describe())

print("\nZIP prefixes with fewer than 5 digits:")
print(
    customers[
        customers["customer_zip_code_prefix"].astype(str).str.len() < 5
    ]["customer_zip_code_prefix"].head(10)
)

# 6. Check unique cities and states
print("\nUnique cities:", customers["customer_city"].nunique())
print("Unique states:", customers["customer_state"].nunique())

print("\nCustomer state counts:")
print(customers["customer_state"].value_counts())

# 7. Check blank city and state values
print("\nBlank cities:",
      customers["customer_city"].str.strip().eq("").sum())

print("Blank states:",
      customers["customer_state"].str.strip().eq("").sum())
```

```python

# Create a copy for cleaning
customers_cleaned = customers.copy()

# 1. Convert ZIP code prefixes to strings
# Ensure all prefixes contain five digits
customers_cleaned["customer_zip_code_prefix"] = (
    customers_cleaned["customer_zip_code_prefix"]
    .astype(str)
    .str.zfill(5)
)

# 2. Remove unnecessary leading and trailing spaces
customers_cleaned["customer_city"] = (
    customers_cleaned["customer_city"].str.strip()
)

customers_cleaned["customer_state"] = (
    customers_cleaned["customer_state"].str.strip().str.upper()
)

# 3. Check the results
print("Original shape:", customers.shape)
print("Cleaned shape:", customers_cleaned.shape)

print("\nUpdated data types:")
print(customers_cleaned.dtypes)

print("\nMissing values:")
print(customers_cleaned.isnull().sum())

print("\nDuplicate customer IDs:",
      customers_cleaned["customer_id"].duplicated().sum())

print("\nDuplicate unique customer IDs:",
      customers_cleaned["customer_unique_id"].duplicated().sum())

print("\nZIP code prefix examples:")
print(customers_cleaned["customer_zip_code_prefix"].head(10))

print("\nPreview:")
print(customers_cleaned.head())
```

```python
customers_cleaned.to_csv("customers_cleaned.csv", index=False)

print("Customers dataset saved successfully!")
```

Customers Dataset – Data Cleaning Summary

The Olist customers dataset originally contained 99,441 rows and 5 columns.

Cleaning Steps

Checked for missing values; none were found.
Checked for fully duplicated rows and duplicate customer_id values; none were found.
Identified 3,345 duplicate customer_unique_id values. These were retained because they represent customers who may have placed multiple orders.
Converted customer_zip_code_prefix from integer to string to preserve its identifier format.
Used .str.zfill(5) to ensure all ZIP code prefixes contain five digits.
Used .str.strip() to remove unnecessary leading and trailing spaces from city and state names.
Standardized state abbreviations to uppercase.
Final Result

The cleaned dataset retains all 99,441 rows and 5 columns. No customer records were deleted. The cleaned version is saved as customers_cleaned.csv, while the original dataset remains unchanged.

Relevance to Research

This dataset will be used to connect customer information with orders and reviews. It can support the analysis of customer satisfaction (Research Questions 1–4) and help identify repeat customers using customer_unique_id.
