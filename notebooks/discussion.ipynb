# Chapter 5 – Discussion

## 5.1 Introduction

This chapter discusses the empirical findings presented in Chapter 4
in relation to the research questions, relevant academic literature,
and the theoretical foundations of the study.

The discussion focuses on the relationships between delivery
performance, delivery speed, freight costs, product categories,
marketing funnel conversion, and the marketplace performance
of acquired sellers.

The findings are interpreted to understand their implications
for customer satisfaction and marketing performance in
e-commerce.

Furthermore, this chapter considers the theoretical and managerial
implications of the results, as well as the limitations of the study.

## 5.2 Discussion of RQ1 – Delivery Performance

The findings indicate a clear relationship between delivery performance
and customer satisfaction. Orders delivered late received substantially
lower review scores than orders delivered on time or earlier than
estimated.

The average review score for late deliveries was 2.57, compared with
4.30 for orders delivered on time or early. The Mann–Whitney U test
confirmed that this difference was statistically significant
(U = 148,395,758, p < .001). The rank-biserial correlation of
approximately -0.55 also indicates a substantial difference between
the two delivery groups.

These findings suggest that delivery reliability is an important
aspect of the customer experience in e-commerce. Customers may perceive
late delivery as a failure to meet expectations, particularly when
the estimated delivery date has already been communicated during the
purchase process.

From a managerial perspective, the results highlight the importance
of accurate delivery estimates, reliable logistics processes, and
proactive communication when delays occur. E-commerce businesses
should monitor late delivery rates and identify operational problems
that may negatively affect customer satisfaction.

However, the results demonstrate an association rather than a causal
relationship. Other factors, such as product quality, seller
performance, and customer expectations, may also influence review
scores.

Overall, the findings support the alternative hypothesis (H1) for RQ1
and indicate that late delivery is associated with lower customer
satisfaction.

## 5.3 Discussion of RQ2 – Delivery Speed

The results reveal a negative relationship between delivery duration
and customer satisfaction. The Spearman correlation analysis showed
a statistically significant negative association between the number
of delivery days and the average review score
(ρ = -0.2353, p < .001).

This indicates that longer delivery times tend to be associated with
lower customer review scores. Although the relationship is not
particularly strong, it suggests that delivery speed is a relevant
factor in the overall customer experience.

The descriptive results provide further insight into this relationship.
Orders with a review score of 1 had an average delivery duration of
21.39 days, whereas orders with a review score of 5 had an average
delivery duration of 10.68 days. This difference illustrates the
potential importance of timely delivery from the customer's perspective.

The sensitivity analysis also showed that the negative association
remained when orders with delivery durations exceeding 30 or 60 days
were excluded. Therefore, the observed relationship is not explained
solely by a small number of extremely delayed orders.

From a managerial perspective, e-commerce businesses should monitor
delivery duration and identify opportunities to improve fulfilment
and logistics efficiency. Reducing unnecessary delays and providing
realistic delivery estimates may contribute to a better customer
experience.

However, delivery speed should not be considered in isolation.
Customer satisfaction may also depend on product quality, price,
communication, and the accuracy of delivery expectations.

Overall, the findings support the alternative hypothesis (H1) for
RQ2. Longer delivery durations are associated with lower customer
satisfaction, although the correlation does not establish causation.

## 5.4 Discussion of RQ3 – Freight Costs

The findings indicate a statistically significant but very weak
negative association between total freight costs and customer
satisfaction. The Spearman correlation coefficient was -0.0881
(p < .001).

This suggests that orders with higher freight costs tend to receive
slightly lower review scores. However, the strength of this
relationship is limited. The large sample size of 95,178 orders
means that even a very weak association can be statistically
significant.

The comparison of freight cost quartiles provides additional
insight. Orders in the lowest freight cost quartile had an average
review score of 4.32, while orders in the highest freight cost
quartile had an average score of 3.94. This difference suggests
that customers may be less satisfied when they pay higher delivery
costs.

One possible explanation is that customers evaluate delivery costs
in relation to the value and expectations of their purchases.
High freight costs may be perceived as unfavourable when customers
do not believe that the delivery service provides sufficient value.

From a managerial perspective, e-commerce businesses should consider
transparent shipping prices and communicate delivery costs clearly
before checkout. They may also investigate whether free-shipping
thresholds or more flexible delivery options could improve the
customer experience.

Nevertheless, freight costs should not be interpreted independently
of other order characteristics. Larger or heavier orders, for example,
may naturally generate higher freight costs. Product value, order size,
destination, and delivery distance may also influence customer
satisfaction.

Overall, the findings support the alternative hypothesis (H1) for
RQ3, but the relationship is very weak. Freight costs appear to be
a comparatively minor factor in customer satisfaction within the
analysed dataset. The results demonstrate association, not causation.

## 5.5 Discussion of RQ4 – Product Categories

The findings indicate that customer satisfaction differs significantly
across product categories. The Kruskal–Wallis test showed statistically
significant differences in review score distributions across the
59 product categories analysed (H = 817.70, p < .001).

The descriptive results showed that some categories received
considerably higher average review scores than others. For example,
books general interest had an average review score of approximately
4.53, while office furniture had an average score of approximately
3.64. This difference suggests that the customer experience may vary
depending on the type of product purchased.

One possible explanation is that customer expectations differ
between product categories. Customers may evaluate product quality,
functionality, packaging, and suitability differently depending on
the product. Categories involving furniture or larger products may
also involve more complex delivery and handling requirements.

The sensitivity analysis, which excluded orders containing products
from multiple categories, produced a similar result
(H = 778.40, p < .001). This supports the robustness of the overall
finding that review score distributions differ across categories.

From a managerial perspective, e-commerce businesses should avoid
assuming that customer satisfaction is consistent across all product
categories. Category-specific monitoring of customer reviews,
product quality, seller performance, and delivery processes may help
identify areas requiring improvement.

However, the results should be interpreted with methodological
caution. Some orders contained products from more than one category.
Although the sensitivity analysis produced consistent findings,
the initial category-level observations are not fully independent.
Furthermore, the Kruskal–Wallis test identifies differences in
distributions but does not establish which operational factors
explain those differences.

Overall, the findings support the alternative hypothesis (H1) for
RQ4. Customer review score distributions differ significantly
between product categories, although the results do not establish
that product category itself causes differences in satisfaction.

## 5.6 Discussion of RQ5 – Marketing Funnel Conversion Performance

The findings reveal a statistically significant association between
lead origin and conversion into an acquired seller. The chi-square
test of independence showed that conversion outcomes differed
across lead origins (χ²(9) = 137.687, p < .001).

The overall conversion rate was 10.53%, with 842 converted leads
out of 8,000 potential sellers. Conversion rates varied considerably
between lead origins. Among the identified marketing channels,
paid search achieved the highest conversion rate at 12.30%,
followed by organic search at 11.80%. Social media had a lower
conversion rate of 5.56%, while email recorded 3.04%.

Although organic search generated the largest number of converted
leads (271), paid search achieved a slightly higher conversion rate.
This distinction is important because the number of conversions
depends not only on channel effectiveness but also on the number
of leads generated by each channel.

The chi-square test indicates that lead origin and conversion
are not independent. However, Cramér's V of 0.131 suggests that
the strength of this association is weak. Therefore, lead origin
is relevant to conversion performance, but it is unlikely to be
the only factor determining whether a potential seller becomes
an acquired seller.

From a managerial perspective, the findings suggest that marketing
teams should evaluate channels using both conversion rates and
the absolute number of acquired sellers. Paid search may be useful
for generating relatively high conversion rates, while organic
search contributes a larger volume of acquired sellers. Marketing
resources should therefore be allocated with consideration of
both channel efficiency and acquisition volume.

An important limitation concerns the category labelled
"unknown", which achieved the highest conversion rate at 16.65%.
Because this group does not identify a specific marketing channel,
it should not be interpreted as evidence of a particularly
effective marketing activity.

Overall, the findings support the alternative hypothesis (H1)
for RQ5. Lead origin is significantly associated with seller
conversion, although the strength of the association is weak.
The analysis does not establish that a particular marketing
channel causes higher conversion.

## 5.7 Discussion of RQ6 – Acquired Seller Performance

The findings provide an overview of the marketplace performance
of sellers acquired through the Olist marketing funnel. Out of
842 acquired sellers, 380 (45.13%) could be matched to marketplace
seller records. These matched sellers generated 4,518 marketplace
orders and sold 5,044 items, with a total product value of
approximately R$676,851.46.

The order-weighted average review score across 4,405 eligible
delivered orders was 4.28 out of 5. The corresponding
order-weighted average delivery duration was 9.79 days, while
the late delivery rate was 7.24%. These results suggest that
the matched acquired sellers achieved generally positive
customer satisfaction and delivery performance within the
observed marketplace data.

However, performance varied across business segments. Sports
and leisure sellers recorded an average review score of 4.70
and a late delivery rate of 1.62%. Health and beauty sellers
also demonstrated relatively strong performance, with an
average review score of 4.58 and a late delivery rate of 4.87%.

In contrast, some segments showed potential areas for improvement.
Watches had a late delivery rate of 16.84%, based on 580 orders
from three matched sellers. Home decor recorded a late delivery
rate of 13.78% across 361 orders. These differences suggest
that operational performance may vary across business segments.

From a managerial perspective, the findings highlight the
importance of monitoring acquired sellers after recruitment.
Marketing acquisition alone does not guarantee strong marketplace
performance. Seller onboarding, logistics support, service
quality monitoring, and category-specific assistance may help
new sellers establish sustainable marketplace operations.

Nevertheless, the results must be interpreted cautiously.
Only 380 of the 842 acquired sellers could be matched to
marketplace seller records. The remaining 462 sellers had
no linked marketplace orders in the analysed data. This does
not necessarily mean that they were inactive or unsuccessful;
it means that their marketplace performance could not be
measured using the available matching data.

Furthermore, the analysis is descriptive and does not include
an inferential statistical test. Therefore, the results provide
an overview of observed seller performance but do not establish
that marketing acquisition caused positive marketplace outcomes.

Overall, the findings suggest that matched acquired sellers
demonstrated generally positive customer satisfaction and
delivery performance, although differences between business
segments and the incomplete seller matching limit the conclusions.

## 5.8 Theoretical Implications

The findings contribute to the understanding of customer satisfaction
and marketing performance in e-commerce by demonstrating how
operational and marketing-related factors are associated with
marketplace outcomes.

First, the results highlight the importance of delivery reliability.
The substantial difference in review scores between late and
on-time or early deliveries suggests that meeting delivery
expectations is an important element of the customer experience.
This supports the view that fulfilment performance should be
considered an important dimension of perceived service quality.

Second, the negative relationship between delivery duration and
customer satisfaction provides further evidence that delivery
speed is relevant to the post-purchase experience. However, the
relationship is weaker than the difference observed between
late and on-time deliveries. This suggests that customers may
respond not only to the number of days required for delivery
but also to whether the delivery meets the promised timeframe.

Third, the freight cost analysis indicates that price-related
delivery factors are associated with customer satisfaction,
although the observed relationship is very weak. This suggests
that freight costs may form part of the overall value perception
rather than acting as a dominant factor in isolation.

Fourth, the differences in review score distributions across
product categories indicate that customer satisfaction is not
uniform across the marketplace. Product characteristics and
category-specific expectations may contribute to variations
in the customer experience.

Fifth, the marketing funnel analysis demonstrates that lead
origin is associated with seller conversion. Nevertheless,
the weak strength of this association suggests that acquisition
channel alone provides only limited insight into conversion
performance. Other factors, such as lead characteristics,
marketing activities, and seller suitability, may also be relevant.

Finally, the acquired seller analysis connects marketing
acquisition with subsequent marketplace activity. It shows
the importance of considering not only the number of acquired
sellers but also their observable sales activity, customer
satisfaction, and delivery performance.

Taken together, these findings suggest that customer satisfaction
and marketing performance in e-commerce should be understood
as interconnected but distinct dimensions. Marketing activities
may contribute to seller acquisition, while operational
performance influences the customer experience after a purchase.
The results therefore support an integrated perspective that
considers both acquisition activities and post-purchase service
performance.

## 5.9 Managerial Implications

The findings provide several practical implications for e-commerce
businesses and marketplace operators.

**1. Improve delivery reliability**

The substantial difference in customer review scores between
late and on-time or early deliveries highlights the importance
of reliable fulfilment. Businesses should monitor late delivery
rates, improve coordination with logistics providers, and identify
the main causes of delivery delays. Accurate delivery estimates
and proactive communication may also help manage customer
expectations.

**2. Monitor delivery speed**

The negative association between delivery duration and review
scores suggests that businesses should continuously monitor
the time required to fulfil and deliver orders. Improvements
in warehouse processes, order handling, and logistics planning
may help reduce unnecessary delays.

**3. Maintain transparent freight pricing**

Although the relationship between freight costs and customer
satisfaction was weak, customers may still consider shipping
costs when evaluating the overall value of a purchase.
Businesses should communicate freight charges clearly before
checkout and consider flexible delivery options or free-shipping
thresholds where commercially appropriate.

**4. Apply category-specific performance management**

The differences in review scores across product categories
suggest that a uniform management approach may not be sufficient.
Marketplace operators should monitor customer feedback and
operational indicators at category level. Categories with
lower satisfaction levels may require closer investigation
of product quality, seller service, packaging, and delivery
processes.

**5. Evaluate marketing channels using multiple KPIs**

The marketing funnel findings show that conversion rates
and conversion volumes provide different insights into
channel performance. Marketing teams should monitor both
measures rather than relying exclusively on the number of
acquired sellers or the conversion rate. Further evaluation
of acquisition costs and the long-term performance of acquired
sellers would provide a more complete assessment of marketing
effectiveness.

**6. Monitor acquired sellers after onboarding**

The acquired seller analysis indicates that seller acquisition
should not be treated as the final measure of marketing success.
Marketplace operators should also monitor seller activity,
order volume, customer reviews, and delivery performance.
Targeted onboarding and operational support may help sellers
establish sustainable marketplace activity.

Overall, the findings suggest that e-commerce businesses
should adopt an integrated performance management approach.
Marketing acquisition, seller development, logistics reliability,
and customer satisfaction should be monitored together
to support sustainable marketplace performance.

## 5.10 Limitations of the Study

Despite providing useful insights into customer satisfaction
and marketing performance in e-commerce, this study has several
limitations that should be considered when interpreting the
findings.

**1. Limitations of the dataset**

The analysis is based on the Olist Brazilian E-Commerce and
Marketing Funnel datasets. These datasets represent a specific
marketplace and business context. Therefore, the findings
may not be directly generalisable to other e-commerce
platforms, countries, or market environments.

**2. Observational nature of the analysis**

The study uses observational data rather than an experimental
research design. Consequently, the statistical relationships
identified between delivery performance, freight costs,
product categories, marketing lead origins, and customer
satisfaction cannot be interpreted as causal effects.
Other variables may influence the observed relationships.

**3. Measurement of customer satisfaction**

Customer satisfaction is represented by review scores.
Although review scores provide a measurable indicator of
customer feedback, they may not capture the full complexity
of customer satisfaction. Other aspects, such as customer
expectations, perceived value, communication quality, and
the overall shopping experience, are not fully represented
by this measure.

**4. Delivery analysis limitations**

The delivery analyses focus on delivered orders with valid
chronology and available review scores. Orders with missing
or invalid information were excluded from these analyses.
Therefore, the findings apply to the eligible sample rather
than necessarily representing every order in the original
dataset.

**5. Freight cost limitations**

The freight analysis uses total freight value per order.
However, freight costs may depend on factors such as product
weight, order size, delivery distance, and destination.
These factors were not controlled for in the correlation
analysis, which limits the interpretation of the relationship
between freight costs and customer satisfaction.

**6. Product category limitations**

Some orders contained products from multiple categories.
This resulted in multiple category assignments for certain
orders in the initial category analysis. A sensitivity
analysis using only single-category orders produced similar
results, but the category-level observations should still
be interpreted with methodological caution.

**7. Marketing funnel limitations**

The marketing funnel analysis examines the association
between lead origin and conversion. However, it does not
include a complete assessment of marketing expenditure,
customer acquisition cost, or the long-term value of acquired
sellers. Therefore, conversion rates alone cannot establish
the overall financial effectiveness of individual marketing
channels.

**8. Acquired seller matching limitations**

Only 380 of the 842 acquired sellers could be matched to
marketplace seller records. Consequently, the seller
performance analysis represents matched sellers rather
than the entire acquired seller population. The remaining
sellers could not be evaluated using the available
marketplace performance data.

Additionally, differences in acquisition dates and the
available observation period may have affected the time
sellers had to generate marketplace activity.

**9. Statistical analysis limitations**

The study applies non-parametric statistical tests and
descriptive methods to examine the research questions.
Although these methods are appropriate for the analyses
conducted, they do not control for all potential confounding
variables. Furthermore, statistically significant findings
do not necessarily indicate strong or practically important
relationships.

Future research could address these limitations by using
additional datasets, controlling for relevant explanatory
variables, examining longer observation periods, and
including more comprehensive measures of customer and
seller performance.


## 5.11 Chapter Summary

This chapter discussed the empirical findings in relation to
the six research questions concerning customer satisfaction
and marketing performance in e-commerce.

The discussion highlighted the importance of delivery
reliability and delivery speed. Late deliveries were associated
with substantially lower customer review scores, while longer
delivery durations were also associated with lower customer
satisfaction.

Freight costs showed a statistically significant but very
weak negative association with review scores. In addition,
customer satisfaction differed significantly across product
categories, suggesting that the customer experience varies
across different types of products.

The marketing funnel analysis demonstrated a significant
association between lead origin and seller conversion,
although the strength of this relationship was weak.
The acquired seller analysis further showed that matched
sellers generated marketplace activity and generally
achieved positive customer review scores and delivery
performance, while performance varied across business
segments.

The chapter also considered the theoretical and managerial
implications of these findings. It highlighted the importance
of integrating marketing acquisition, seller development,
logistics performance, and customer satisfaction into
marketplace management.

Finally, the limitations of the study were discussed,
including the observational nature of the data, the
restricted measurement of customer satisfaction, and
the incomplete matching of acquired sellers to marketplace
records.
